```python
SIGNATURES = {
    b"\xFF\xD8\xFF": "JPEG image",
    b"\x89PNG": "PNG image",
    b"%PDF": "PDF document",
    b"PK\x03\x04": "ZIP archive (or .docx/.xlsx)",
}


# Function to identify file type
def identify_file(path):
    with open(path, "rb") as f:
        header = f.read(8)

    for sig, filetype in SIGNATURES.items():
        if header.startswith(sig):
            return filetype

    return "Unknown file type"


# Create sample files with fake/renamed extensions
# Testing signature detection

# Simulate a JPEG file renamed as .txt
with open("photo.txt", "wb") as f:
    f.write(b"\xFF\xD8\xFF\xE0" + b"\x00" * 20)

# Simulate a PDF file renamed as .dat
with open("document.dat", "wb") as f:
    f.write(b"%PDF-1.4" + b"\x00" * 20)


# Identify the actual file types
for filename in ["photo.txt", "document.dat"]:
    print(f"{filename:15s} -> Actual type: {identify_file(filename)}")
```

```python
#EXP 10. Write a Python program to parse raw email headers and trace indicators of a spoofed sender.

raw_header = """Delivered-To: victim@example.com
Received: from mail-relay-99.suspicious-host.ru (unknown [45.33.32.156])
by mx.example.com; Tue, 08 Jul 2026 09:15:00 +0000
From: "Bank Support"
Reply-To: attacker@totallynotscam.net
Subject: Urgent: Verify your account
"""


def analyze_header(header_text):
    findings = []

    for line in header_text.splitlines():

        # Check for suspicious relay server
        if line.startswith("Received:") and "suspicious" in line.lower():
            findings.append(
                f"Suspicious relay server found: {line.strip()}"
            )

        # Check for Reply-To header
        if line.startswith("Reply-To:"):
            findings.append(
                f"Reply-To differs from sender - possible spoofing: {line.strip()}"
            )

        # Check for suspicious sender domain
        if line.startswith("From:") and "-secure-verify" in line:
            findings.append(
                f"Sender domain uses suspicious naming: {line.strip()}"
            )

    return findings


print("Email Header Forensic Analysis")

for finding in analyze_header(raw_header):
    print(" -", finding)
```

```python
# EXP 14: Digital Evidence Collection & Cataloguing

import os
import hashlib

# Create evidence directory
evidence_dir = "digital_evidence"
os.makedirs(evidence_dir, exist_ok=True)

# Create sample digital evidence
sample_evidence = {
    "email_log.txt": """From: attacker@mail.com
To: victim@mail.com
Subject: Invoice
Please pay immediately.""",

    "browser_history.txt": """http://malicious-site.com/login
http://bank.com/transfer""",

    "chat_message.txt": """Hey, did you send the file?""",

    "system_log.txt": """2026-07-24 10:15:32 - USB device connected: E:\\"""
}

# Write sample evidence files
for filename, content in sample_evidence.items():
    with open(os.path.join(evidence_dir, filename), "w") as f:
        f.write(content)

# Function to catalog evidence
def catalog_evidence(folder):
    catalog = []

    for fname in sorted(os.listdir(folder)):
        path = os.path.join(folder, fname)

        # Get file metadata
        stat = os.stat(path)

        # Read file contents
        with open(path, "rb") as f:
            data = f.read()

        # Store evidence information
        catalog.append({
            "filename": fname,
            "size_bytes": stat.st_size,
            "md5": hashlib.md5(data).hexdigest()
        })

    return catalog

# Generate evidence catalog
evidence_catalog = catalog_evidence(evidence_dir)

# Display evidence catalog
print("=== DIGITAL EVIDENCE CATALOG ===")

for item in evidence_catalog:
    print("-" * 40)
    print("Filename   :", item["filename"])
    print("Size (Bytes):", item["size_bytes"])
    print("MD5 Hash   :", item["md5"])
```

```python
#EXP15.Volatile vs Non-Volatile Evidence Capture

import os
import psutil
# Evidence directory
evidence_dir = "digital_evidence"
# Create the folder if it does not already exist
os.makedirs(evidence_dir, exist_ok=True)
# Capture volatile evidence
def capture_volatile_evidence():
    processes = [
        p.info
        for p in psutil.process_iter(['pid', 'name'])
    ]
    return {
        "running_process_count": len(processes),
        "sample_processes": processes[:5],
    }
# Capture non-volatile evidence
def capture_nonvolatile_evidence(folder):
    files = sorted(os.listdir(folder))
    return {
        "disk_files": files,
        "file_count": len(files)
    }
# Capture evidence
volatile_snapshot = capture_volatile_evidence()
nonvolatile_snapshot = capture_nonvolatile_evidence(evidence_dir)
# Display results
print("=== DIGITAL FORENSIC EVIDENCE ===")
print(
    "Volatile evidence -> running processes:",
    volatile_snapshot["running_process_count"]
)
print(
    "Sample running processes:",
    volatile_snapshot["sample_processes"]
)
print(
    "Non-volatile evidence -> disk files:",
    nonvolatile_snapshot
)
```

```python
#EXP16.Physical, Logical and Sparse Acquisition

import os
os.makedirs("acquisition_demo/source", exist_ok=True)
source = "acquisition_demo/source"

with open(os.path.join(source, "report.docx"), "w") as f:
    f.write("Confidential quarterly report.")

with open(os.path.join(source, "photo.jpg"), "w") as f:
    f.write("FAKEJPEGDATA")

raw_disk = b"REPORTDOCX_CONTENT" + b"\x00" * 20 + b"DELETED_INVOICE_DATA" + b"\x00" * 10

def physical_acquisition(raw_bytes):
    return bytes(raw_bytes)  # every bit: used + unused + deleted

def logical_acquisition(folder):
    return {f: open(os.path.join(folder, f), "rb").read() for f in os.listdir(folder)}

def sparse_acquisition(folder, targets):
    return {f: open(os.path.join(folder, f), "rb").read() for f in targets if f in os.listdir(folder)}

physical_image = physical_acquisition(raw_disk)
logical_image = logical_acquisition(source)
sparse_image = sparse_acquisition(source, ["report.docx"])

print("Physical image size:", len(physical_image), "bytes")
print("Logical image files:", list(logical_image.keys()))
print("Sparse image files:", list(sparse_image.keys()))
```

```python
#EXP 20 Hashing & the Avalanche Effect

import hashlib

def compute_hashes(data: bytes):
    return {
        "MD5": hashlib.md5(data).hexdigest(),
        "SHA1": hashlib.sha1(data).hexdigest(),
        "SHA256": hashlib.sha256(data).hexdigest(),
    }

original_data = b"hello"
tampered_data = b"Hello"

hashes_original = compute_hashes(original_data)
hashes_tampered = compute_hashes(tampered_data)

print("hello ->", hashes_original)
print("Hello ->", hashes_tampered)
```

```python
#EXP 21 File Carving (Data Recovery)

import os
def make_fake_jpeg(payload: bytes):
    return b"\xff\xd8\xff" + payload + b"\xff\xd9"
jpeg1 = make_fake_jpeg(b"PHOTO_OF_SUSPECT_CAR")
jpeg2 = make_fake_jpeg(b"CCTV_FRAME_CAPTURE")
raw_disk_blob = (
    os.urandom(30) + jpeg1 +
    os.urandom(40) + jpeg2 +
    os.urandom(20)
)
def carve_jpegs(blob: bytes):
    recovered = []
    start_marker = b"\xff\xd8\xff"
    end_marker = b"\xff\xd9"
    pos = 0
    while True:
        start = blob.find(start_marker, pos)
        if start == -1:
            break
        end = blob.find(end_marker, start)
        if end == -1:
            break
        end += len(end_marker)
        recovered.append(blob[start:end])
        pos = end
    return recovered
recovered_files = carve_jpegs(raw_disk_blob)
print("Recovered", len(recovered_files), "file(s) via carving")
```

```python
#EXP 22 File System Feature Simulation (FAT vs NTFS vs EXT)

class FATFileEntry:
    def __init__(self, name, size):
        self.name = name
        self.size = size
class NTFSFileEntry:
    def __init__(self, name, size, permissions="rw-r--r--"):
        self.name = name
        self.size = size
        self.permissions = permissions
        self.journal_entry = f"WRITE {name} {size} bytes"
class EXTFileEntry:
    def __init__(self, name, size, permissions="rw-r--r--"):
        self.name = name
        self.size = size
        self.permissions = permissions
        self.journal_entry = f"WRITE {name} {size} bytes"
fat_file = FATFileEntry("data.txt", 1024)
ntfs_file = NTFSFileEntry("data.txt", 1024)
ext_file = EXTFileEntry("data.txt", 1024)
print("FAT has permissions attribute:", hasattr(fat_file, "permissions"))
print("NTFS has permissions attribute:", hasattr(ntfs_file, "permissions"))
print("EXT has permissions attribute:", hasattr(ext_file, "permissions"))
```

```python
#EXP 23 Slack Space Calculation

CLUSTER_SIZE = 4096  # Typical disk cluster size in bytes
def calculate_slack_space(file_size, cluster_size=CLUSTER_SIZE):
    clusters_needed = -(-file_size // cluster_size)
    allocated_space = clusters_needed * cluster_size
    slack_space = allocated_space - file_size
    return allocated_space, slack_space
file_size = 5000  # bytes
allocated, slack = calculate_slack_space(file_size)
print(
    f"File size: {file_size} bytes -> "
    f"allocated: {allocated} bytes, "
    f"slack space: {slack} bytes"
)

```

```python
#EXP 25  Phishing Email Analysis

import re
from urllib.parse import urlparse
from collections import Counter

# Sample phishing emails
emails = [
    {
        "sender": "support@paypal-secure-verify.com",
        "subject": "URGENT: Verify your account",
        "body": "Click here: http://192.168.10.5/login to verify your account immediately."
    },
    {
        "sender": "security@bank-update-login.com",
        "subject": "Your account is suspended",
        "body": "Your account is suspended. Visit https://bank-update-login.com/verify now."
    },
    {
        "sender": "support@company.com",
        "subject": "Monthly Report",
        "body": "Please find the monthly report attached."
    }
]


# Analyze one email
def analyze_email(email):
    sender = email["sender"]
    subject = email["subject"]
    body = email["body"]

    indicators = []
    urls = []

    # Extract URLs
    urls = re.findall(r"https?://[^\s]+", body)

    # Check urgency words
    if re.search(r"urgent|immediately|suspended|verify", subject + " " + body, re.I):
        indicators.append("Urgency or pressure language")

    # Check URLs
    for url in urls:
        domain = urlparse(url).netloc

        # Check for IP address URL
        if re.match(r"^\d+\.\d+\.\d+\.\d+$", domain):
            indicators.append("URL uses raw IP address")

        # Check suspicious domain keywords
        if any(word in domain.lower() for word in ["secure", "verify", "update", "login"]):
            indicators.append("Suspicious domain naming")

    # Check sender domain
    sender_domain = sender.split("@")[-1]

    if any(word in sender_domain.lower() for word in ["secure", "verify", "update", "login"]):
        indicators.append("Suspicious sender domain")

    # Calculate risk score
    risk_score = len(indicators) * 25

    if risk_score >= 50:
        classification = "PHISHING"
    elif risk_score > 0:
        classification = "SUSPICIOUS"
    else:
        classification = "LEGITIMATE"

    return {
        "sender": sender,
        "subject": subject,
        "urls": urls,
        "indicators": indicators,
        "risk_score": risk_score,
        "classification": classification
    }


# Analyze all emails
reports = []

for email in emails:
    reports.append(analyze_email(email))


# Display phishing report
print("=== PHISHING EMAIL ANALYSIS REPORT ===")

for i, report in enumerate(reports, 1):
    print(f"\nEmail {i}")
    print("Sender:", report["sender"])
    print("Subject:", report["subject"])
    print("Classification:", report["classification"])
    print("Risk Score:", report["risk_score"], "/ 100")

    print("Extracted URLs:")
    if report["urls"]:
        for url in report["urls"]:
            print(" -", url)
    else:
        print(" - None")

    print("Indicators:")
    if report["indicators"]:
        for indicator in report["indicators"]:
            print(" -", indicator)
    else:
        print(" - None")


# Summary statistics
phishing_count = sum(
    1 for r in reports if r["classification"] == "PHISHING"
)

suspicious_count = sum(
    1 for r in reports if r["classification"] == "SUSPICIOUS"
)

legitimate_count = sum(
    1 for r in reports if r["classification"] == "LEGITIMATE"
)

all_indicators = []

for report in reports:
    all_indicators.extend(report["indicators"])

indicator_count = Counter(all_indicators)

print("\n=== SUMMARY STATISTICS ===")
print("Total Emails:", len(reports))
print("Phishing:", phishing_count)
print("Suspicious:", suspicious_count)
print("Legitimate:", legitimate_count)

print("\nPhishing Tactics Used:")
for tactic, count in indicator_count.items():
    print(f" - {tactic}: {count}")
```

```python
#EXP 26  Malware Hash Analysis

import hashlib
import csv

# Sample files
files = {
    "file1.txt": b"Hello World",
    "file2.txt": b"This is a test file",
    "malware.txt": b"Simulated malware sample"
}

# Known malware SHA-256 hashes (example)
malware_db = set()

# Create hash database
with open("malware_hash_database.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Filename", "MD5", "SHA1", "SHA256", "Detection"])

    for name, data in files.items():
        md5 = hashlib.md5(data).hexdigest()
        sha1 = hashlib.sha1(data).hexdigest()
        sha256 = hashlib.sha256(data).hexdigest()

        # Simulate malware database
        if name == "malware.txt":
            malware_db.add(sha256)

        result = "MALWARE" if sha256 in malware_db else "SAFE"

        writer.writerow([name, md5, sha1, sha256, result])

        print(name, "->", result)

print("\nDatabase saved as malware_hash_database.csv")
```

```python
# EXP 28: Hacking Attack Simulation (Ethical)

import re

# Simulated attack requests
requests = [
    "login user=admin password=123456",
    "login user=admin password=password",
    "login user=admin password=admin123",
    "search?q=' OR '1'='1",
    "search?q=hello",
    "file=../../etc/passwd"
]

# Attack signatures
patterns = {
    "Weak Password": r"password=(123456|password|admin123)",
    "SQL Injection": r"('|--|OR\s+['\"]?1['\"]?\s*=\s*['\"]?1)",
    "Directory Traversal": r"\.\./|\.\.\\"
}

print("=== ATTACK PATTERN ANALYSIS ===")

for request in requests:
    detected = []

    for attack, pattern in patterns.items():
        if re.search(pattern, request, re.IGNORECASE):
            detected.append(attack)

    print("\nRequest:", request)

    if detected:
        print("Detected:", ", ".join(detected))
    else:
        print("Status: Normal")

print("\n=== WAF RULE SUGGESTIONS ===")
print("1. Rate-limit repeated login attempts.")
print("2. Use parameterized SQL queries.")
print("3. Block ../ and ..\\ path traversal patterns.")
print("4. Enforce strong password policies.")
```
