```python
import hashlib

# Function to calculate SHA-256 hash
def sha256_of(filename):
    with open(filename, "rb") as f:
        return hashlib.sha256(f.read()).hexdigest()


# Create a sample "disk" file
# This simulates a small storage device
with open("original_disk.img", "wb") as f:
    f.write(b"HEADER" + bytes(range(256)) * 4 + b"FOOTER")


# Step 1: Create a bit-for-bit forensic copy
with open("original_disk.img", "rb") as src:
    with open("forensic_copy.img", "wb") as dst:
        dst.write(src.read())


# Step 2: Verify the copy using SHA-256 hashing
original_hash = sha256_of("original_disk.img")
copy_hash = sha256_of("forensic_copy.img")


# Display the results
print("Original image hash:", original_hash)
print("Forensic copy hash :", copy_hash)

if original_hash == copy_hash:
    print("Match: VERIFIED - exact bit-stream copy")
else:
    print("MISMATCH - copy corrupted")
```

```python
# ==========================================================
# EXPERIMENT - 1
# Experiment Name: Windows Event Log Brute-Force Login Detector
# Topic: Windows Forensics (Event Logs - Event ID 4624 & 4625)
#
# Aim:
# Detect brute-force login attempts by identifying multiple
# failed logins followed by a successful login.
# ==========================================================

from datetime import datetime, timedelta

# ----------------------------------------------------------
# Function to Parse Timestamp
# ----------------------------------------------------------

def parse_time(t):
    return datetime.strptime(t, "%Y-%m-%d %H:%M:%S")


# ----------------------------------------------------------
# Function to Detect Brute Force Attack
# ----------------------------------------------------------

def detect_bruteforce(events, threshold=5, window_minutes=2):

    events = sorted(events, key=lambda e: parse_time(e["timestamp"]))

    by_account = {}

    for e in events:
        by_account.setdefault(e["account"], []).append(e)

    results = {}

    for account, acc_events in by_account.items():

        failures = [e for e in acc_events if e["event_id"] == 4625]
        successes = [e for e in acc_events if e["event_id"] == 4624]

        flagged = False

        for i in range(len(failures)):

            window_start = parse_time(failures[i]["timestamp"])
            window_end = window_start + timedelta(minutes=window_minutes)

            count = sum(
                1
                for f in failures
                if window_start <= parse_time(f["timestamp"]) <= window_end
            )

            if count >= threshold:
                flagged = True
                break

        if flagged:

            followed_by_success = any(
                parse_time(s["timestamp"]) >
                parse_time(failures[-1]["timestamp"])
                for s in successes
            )

            results[account] = {
                "failed_attempts": len(failures),
                "followed_by_success": followed_by_success,
                "source_ips": sorted(
                    {f["source_ip"] for f in failures}
                ),
            }

    return results


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment1():

    events = []

    base = datetime(2026, 1, 15, 3, 40, 0)

    for i in range(6):
        events.append({
            "event_id": 4625,
            "account": "Administrator",
            "timestamp": (base + timedelta(seconds=15*i)).strftime("%Y-%m-%d %H:%M:%S"),
            "source_ip": "203.0.113.7",
        })

    events.append({
        "event_id": 4624,
        "account": "Administrator",
        "timestamp": (base + timedelta(seconds=100)).strftime("%Y-%m-%d %H:%M:%S"),
        "source_ip": "203.0.113.7",
    })

    events.append({
        "event_id": 4624,
        "account": "jsmith",
        "timestamp": "2026-01-15 09:00:00",
        "source_ip": "10.0.0.5",
    })

    results = detect_bruteforce(events)

    assert "Administrator" in results
    assert results["Administrator"]["failed_attempts"] == 6
    assert results["Administrator"]["followed_by_success"] is True
    assert results["Administrator"]["source_ips"] == ["203.0.113.7"]
    assert "jsmith" not in results

    print("All test cases passed.")


# ----------------------------------------------------------
# Run Test
# ----------------------------------------------------------

test_experiment1()
```

```python
SIGNATURES = {
    b"\xFF\xD8\xFF": "JPEG image",
    b"\x89PNG": "PNG image",
    b"%PDF": "PDF document",
    b"PK\x03\x04": "ZIP archive (or .docx/.xlsx)",
}


# Function to identify file type
def identify_file(path):
    with open(path, "rb") as f:
        header = f.read(8)

    for sig, filetype in SIGNATURES.items():
        if header.startswith(sig):
            return filetype

    return "Unknown file type"


# Create sample files with fake/renamed extensions
# Testing signature detection

# Simulate a JPEG file renamed as .txt
with open("photo.txt", "wb") as f:
    f.write(b"\xFF\xD8\xFF\xE0" + b"\x00" * 20)

# Simulate a PDF file renamed as .dat
with open("document.dat", "wb") as f:
    f.write(b"%PDF-1.4" + b"\x00" * 20)


# Identify the actual file types
for filename in ["photo.txt", "document.dat"]:
    print(f"{filename:15s} -> Actual type: {identify_file(filename)}")
```

```python
#EXP 10. Write a Python program to parse raw email headers and trace indicators of a spoofed sender.

raw_header = """Delivered-To: victim@example.com
Received: from mail-relay-99.suspicious-host.ru (unknown [45.33.32.156])
by mx.example.com; Tue, 08 Jul 2026 09:15:00 +0000
From: "Bank Support"
Reply-To: attacker@totallynotscam.net
Subject: Urgent: Verify your account
"""


def analyze_header(header_text):
    findings = []

    for line in header_text.splitlines():

        # Check for suspicious relay server
        if line.startswith("Received:") and "suspicious" in line.lower():
            findings.append(
                f"Suspicious relay server found: {line.strip()}"
            )

        # Check for Reply-To header
        if line.startswith("Reply-To:"):
            findings.append(
                f"Reply-To differs from sender - possible spoofing: {line.strip()}"
            )

        # Check for suspicious sender domain
        if line.startswith("From:") and "-secure-verify" in line:
            findings.append(
                f"Sender domain uses suspicious naming: {line.strip()}"
            )

    return findings


print("Email Header Forensic Analysis")

for finding in analyze_header(raw_header):
    print(" -", finding)
```

```python
# EXP 14: Digital Evidence Collection & Cataloguing

import os
import hashlib

# Create evidence directory
evidence_dir = "digital_evidence"
os.makedirs(evidence_dir, exist_ok=True)

# Create sample digital evidence
sample_evidence = {
    "email_log.txt": """From: attacker@mail.com
To: victim@mail.com
Subject: Invoice
Please pay immediately.""",

    "browser_history.txt": """http://malicious-site.com/login
http://bank.com/transfer""",

    "chat_message.txt": """Hey, did you send the file?""",

    "system_log.txt": """2026-07-24 10:15:32 - USB device connected: E:\\"""
}

# Write sample evidence files
for filename, content in sample_evidence.items():
    with open(os.path.join(evidence_dir, filename), "w") as f:
        f.write(content)

# Function to catalog evidence
def catalog_evidence(folder):
    catalog = []

    for fname in sorted(os.listdir(folder)):
        path = os.path.join(folder, fname)

        # Get file metadata
        stat = os.stat(path)

        # Read file contents
        with open(path, "rb") as f:
            data = f.read()

        # Store evidence information
        catalog.append({
            "filename": fname,
            "size_bytes": stat.st_size,
            "md5": hashlib.md5(data).hexdigest()
        })

    return catalog

# Generate evidence catalog
evidence_catalog = catalog_evidence(evidence_dir)

# Display evidence catalog
print("=== DIGITAL EVIDENCE CATALOG ===")

for item in evidence_catalog:
    print("-" * 40)
    print("Filename   :", item["filename"])
    print("Size (Bytes):", item["size_bytes"])
    print("MD5 Hash   :", item["md5"])
```

```python
#EXP15.Volatile vs Non-Volatile Evidence Capture

import os
import psutil
# Evidence directory
evidence_dir = "digital_evidence"
# Create the folder if it does not already exist
os.makedirs(evidence_dir, exist_ok=True)
# Capture volatile evidence
def capture_volatile_evidence():
    processes = [
        p.info
        for p in psutil.process_iter(['pid', 'name'])
    ]
    return {
        "running_process_count": len(processes),
        "sample_processes": processes[:5],
    }
# Capture non-volatile evidence
def capture_nonvolatile_evidence(folder):
    files = sorted(os.listdir(folder))
    return {
        "disk_files": files,
        "file_count": len(files)
    }
# Capture evidence
volatile_snapshot = capture_volatile_evidence()
nonvolatile_snapshot = capture_nonvolatile_evidence(evidence_dir)
# Display results
print("=== DIGITAL FORENSIC EVIDENCE ===")
print(
    "Volatile evidence -> running processes:",
    volatile_snapshot["running_process_count"]
)
print(
    "Sample running processes:",
    volatile_snapshot["sample_processes"]
)
print(
    "Non-volatile evidence -> disk files:",
    nonvolatile_snapshot
)
```

```python
#EXP16.Physical, Logical and Sparse Acquisition

import os
os.makedirs("acquisition_demo/source", exist_ok=True)
source = "acquisition_demo/source"

with open(os.path.join(source, "report.docx"), "w") as f:
    f.write("Confidential quarterly report.")

with open(os.path.join(source, "photo.jpg"), "w") as f:
    f.write("FAKEJPEGDATA")

raw_disk = b"REPORTDOCX_CONTENT" + b"\x00" * 20 + b"DELETED_INVOICE_DATA" + b"\x00" * 10

def physical_acquisition(raw_bytes):
    return bytes(raw_bytes)  # every bit: used + unused + deleted

def logical_acquisition(folder):
    return {f: open(os.path.join(folder, f), "rb").read() for f in os.listdir(folder)}

def sparse_acquisition(folder, targets):
    return {f: open(os.path.join(folder, f), "rb").read() for f in targets if f in os.listdir(folder)}

physical_image = physical_acquisition(raw_disk)
logical_image = logical_acquisition(source)
sparse_image = sparse_acquisition(source, ["report.docx"])

print("Physical image size:", len(physical_image), "bytes")
print("Logical image files:", list(logical_image.keys()))
print("Sparse image files:", list(sparse_image.keys()))
```

```python
#EXP 21 File Carving (Data Recovery)

import os
def make_fake_jpeg(payload: bytes):
    return b"\xff\xd8\xff" + payload + b"\xff\xd9"
jpeg1 = make_fake_jpeg(b"PHOTO_OF_SUSPECT_CAR")
jpeg2 = make_fake_jpeg(b"CCTV_FRAME_CAPTURE")
raw_disk_blob = (
    os.urandom(30) + jpeg1 +
    os.urandom(40) + jpeg2 +
    os.urandom(20)
)
def carve_jpegs(blob: bytes):
    recovered = []
    start_marker = b"\xff\xd8\xff"
    end_marker = b"\xff\xd9"
    pos = 0
    while True:
        start = blob.find(start_marker, pos)
        if start == -1:
            break
        end = blob.find(end_marker, start)
        if end == -1:
            break
        end += len(end_marker)
        recovered.append(blob[start:end])
        pos = end
    return recovered
recovered_files = carve_jpegs(raw_disk_blob)
print("Recovered", len(recovered_files), "file(s) via carving")
```

```python
#EXP 23 Slack Space Calculation

CLUSTER_SIZE = 4096  # Typical disk cluster size in bytes
def calculate_slack_space(file_size, cluster_size=CLUSTER_SIZE):
    clusters_needed = -(-file_size // cluster_size)
    allocated_space = clusters_needed * cluster_size
    slack_space = allocated_space - file_size
    return allocated_space, slack_space
file_size = 5000  # bytes
allocated, slack = calculate_slack_space(file_size)
print(
    f"File size: {file_size} bytes -> "
    f"allocated: {allocated} bytes, "
    f"slack space: {slack} bytes"
)

```

```python
#EXP 25  Phishing Email Analysis

import re
from urllib.parse import urlparse
from collections import Counter

# Sample phishing emails
emails = [
    {
        "sender": "support@paypal-secure-verify.com",
        "subject": "URGENT: Verify your account",
        "body": "Click here: http://192.168.10.5/login to verify your account immediately."
    },
    {
        "sender": "security@bank-update-login.com",
        "subject": "Your account is suspended",
        "body": "Your account is suspended. Visit https://bank-update-login.com/verify now."
    },
    {
        "sender": "support@company.com",
        "subject": "Monthly Report",
        "body": "Please find the monthly report attached."
    }
]


# Analyze one email
def analyze_email(email):
    sender = email["sender"]
    subject = email["subject"]
    body = email["body"]

    indicators = []
    urls = []

    # Extract URLs
    urls = re.findall(r"https?://[^\s]+", body)

    # Check urgency words
    if re.search(r"urgent|immediately|suspended|verify", subject + " " + body, re.I):
        indicators.append("Urgency or pressure language")

    # Check URLs
    for url in urls:
        domain = urlparse(url).netloc

        # Check for IP address URL
        if re.match(r"^\d+\.\d+\.\d+\.\d+$", domain):
            indicators.append("URL uses raw IP address")

        # Check suspicious domain keywords
        if any(word in domain.lower() for word in ["secure", "verify", "update", "login"]):
            indicators.append("Suspicious domain naming")

    # Check sender domain
    sender_domain = sender.split("@")[-1]

    if any(word in sender_domain.lower() for word in ["secure", "verify", "update", "login"]):
        indicators.append("Suspicious sender domain")

    # Calculate risk score
    risk_score = len(indicators) * 25

    if risk_score >= 50:
        classification = "PHISHING"
    elif risk_score > 0:
        classification = "SUSPICIOUS"
    else:
        classification = "LEGITIMATE"

    return {
        "sender": sender,
        "subject": subject,
        "urls": urls,
        "indicators": indicators,
        "risk_score": risk_score,
        "classification": classification
    }


# Analyze all emails
reports = []

for email in emails:
    reports.append(analyze_email(email))


# Display phishing report
print("=== PHISHING EMAIL ANALYSIS REPORT ===")

for i, report in enumerate(reports, 1):
    print(f"\nEmail {i}")
    print("Sender:", report["sender"])
    print("Subject:", report["subject"])
    print("Classification:", report["classification"])
    print("Risk Score:", report["risk_score"], "/ 100")

    print("Extracted URLs:")
    if report["urls"]:
        for url in report["urls"]:
            print(" -", url)
    else:
        print(" - None")

    print("Indicators:")
    if report["indicators"]:
        for indicator in report["indicators"]:
            print(" -", indicator)
    else:
        print(" - None")


# Summary statistics
phishing_count = sum(
    1 for r in reports if r["classification"] == "PHISHING"
)

suspicious_count = sum(
    1 for r in reports if r["classification"] == "SUSPICIOUS"
)

legitimate_count = sum(
    1 for r in reports if r["classification"] == "LEGITIMATE"
)

all_indicators = []

for report in reports:
    all_indicators.extend(report["indicators"])

indicator_count = Counter(all_indicators)

print("\n=== SUMMARY STATISTICS ===")
print("Total Emails:", len(reports))
print("Phishing:", phishing_count)
print("Suspicious:", suspicious_count)
print("Legitimate:", legitimate_count)

print("\nPhishing Tactics Used:")
for tactic, count in indicator_count.items():
    print(f" - {tactic}: {count}")
```

```python
from datetime import datetime, timedelta

def parse_time(t):
    return datetime.strptime(t, "%Y-%m-%d %H:%M:%S")

def detect_bruteforce(events, threshold=5, window_minutes=2):
    """
    Detect brute-force login attempts:
    >= threshold failed logons (4625) for the same account
    within window_minutes, followed by a successful logon (4624).
    """

    # Sort events by timestamp
    events = sorted(events, key=lambda e: parse_time(e["timestamp"]))

    # Group events by account
    by_account = {}
    for e in events:
        by_account.setdefault(e["account"], []).append(e)

    results = {}

    # Check each account
    for account, acc_events in by_account.items():

        failures = [e for e in acc_events if e["event_id"] == 4625]
        successes = [e for e in acc_events if e["event_id"] == 4624]

        flagged = False

        # Sliding time window
        for i in range(len(failures)):
            window_start = parse_time(failures[i]["timestamp"])
            window_end = window_start + timedelta(minutes=window_minutes)

            count = sum(
                1
                for f in failures
                if window_start <= parse_time(f["timestamp"]) <= window_end
            )

            if count >= threshold:
                flagged = True

                # Check if a successful login happened after failures
                success_after = any(
                    parse_time(s["timestamp"]) > window_end
                    for s in successes
                )

                results[account] = {
                    "failed_attempts": count,
                    "attack_detected": True,
                    "successful_login_after": success_after,
                }
                break

        if not flagged:
            results[account] = {
                "failed_attempts": len(failures),
                "attack_detected": False,
                "successful_login_after": False,
            }

    return results


# Simulated Event Log
events = [
    {"event_id": 4625, "account": "admin", "timestamp": "2026-08-04 10:00:00", "source_ip": "192.168.1.10"},
    {"event_id": 4625, "account": "admin", "timestamp": "2026-08-04 10:00:20", "source_ip": "192.168.1.10"},
    {"event_id": 4625, "account": "admin", "timestamp": "2026-08-04 10:00:40", "source_ip": "192.168.1.10"},
    {"event_id": 4625, "account": "admin", "timestamp": "2026-08-04 10:01:00", "source_ip": "192.168.1.10"},
    {"event_id": 4625, "account": "admin", "timestamp": "2026-08-04 10:01:20", "source_ip": "192.168.1.10"},
    {"event_id": 4624, "account": "admin", "timestamp": "2026-08-04 10:02:30", "source_ip": "192.168.1.10"},

    {"event_id": 4625, "account": "user1", "timestamp": "2026-08-04 11:00:00", "source_ip": "192.168.1.20"},
    {"event_id": 4625, "account": "user1", "timestamp": "2026-08-04 11:05:00", "source_ip": "192.168.1.20"},
    {"event_id": 4624, "account": "user1", "timestamp": "2026-08-04 11:10:00", "source_ip": "192.168.1.20"},
]

# Run Detection
results = detect_bruteforce(events)

# Display Results
for account, info in results.items():
    print(f"\nAccount: {account}")
    print(f"Failed Attempts: {info['failed_attempts']}")
    print(f"Attack Detected: {info['attack_detected']}")
    print(f"Successful Login After Attack: {info['successful_login_after']}")
```

```python
# ==========================================================
# EXPERIMENT - 2
# Experiment Name: Windows Registry USB Device History Analyzer
# Topic: Windows Forensics (Registry - SYSTEM Hive, USBSTOR Key)
#
# Aim:
# Parse simulated USBSTOR registry data to reconstruct USB
# device history and identify which USB device was connected
# close to a suspected data-exfiltration timestamp.
# ==========================================================

from datetime import datetime

# ----------------------------------------------------------
# Function to Parse USBSTOR Registry Data
# ----------------------------------------------------------

def parse_usbstor(registry_data):
    """
    registry_data:
    {
        device_id: {
            "serial",
            "friendly_name",
            "first_connected",
            "last_connected"
        }
    }

    Returns a list of USB devices sorted by most recent
    last_connected time.
    """

    devices = []

    for device_id, info in registry_data.items():
        entry = {"device_id": device_id}
        entry.update(info)
        devices.append(entry)

    devices.sort(
        key=lambda d: datetime.strptime(
            d["last_connected"], "%Y-%m-%d %H:%M:%S"
        ),
        reverse=True,
    )

    return devices


# ----------------------------------------------------------
# Function to Find Devices Connected Near a Target Time
# ----------------------------------------------------------

def find_device_near_time(devices, target_time_str, window_minutes=30):
    """
    Returns devices whose last_connected timestamp is
    within window_minutes of the target timestamp.
    """

    target = datetime.strptime(target_time_str, "%Y-%m-%d %H:%M:%S")
    matches = []

    for d in devices:
        last = datetime.strptime(
            d["last_connected"], "%Y-%m-%d %H:%M:%S"
        )

        delta_minutes = abs((target - last).total_seconds()) / 60

        if delta_minutes <= window_minutes:
            matches.append(d)

    return matches


# ----------------------------------------------------------
# Simulated Windows Registry (USBSTOR Key)
# ----------------------------------------------------------

registry_data = {
    "USB\\VID_0781&PID_5567\\4C531001234": {
        "serial": "4C531001234",
        "friendly_name": "SanDisk Cruzer Blade",
        "first_connected": "2025-11-01 09:00:00",
        "last_connected": "2026-02-10 17:42:00",
    },

    "USB\\VID_090C&PID_1000\\A1002233": {
        "serial": "A1002233",
        "friendly_name": "Kingston DataTraveler",
        "first_connected": "2024-06-01 10:00:00",
        "last_connected": "2024-06-01 10:15:00",
    },
}


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment2():

    devices = parse_usbstor(registry_data)

    # Check most recent device
    assert devices[0]["friendly_name"] == "SanDisk Cruzer Blade"

    # Search near suspected exfiltration time
    matches = find_device_near_time(
        devices,
        "2026-02-10 17:35:00",
        window_minutes=30
    )

    matched_names = [m["friendly_name"] for m in matches]

    assert "SanDisk Cruzer Blade" in matched_names
    assert "Kingston DataTraveler" not in matched_names

    print("All test cases passed.\n")

    print("===== USB Device History =====")
    for d in devices:
        print("Device Name     :", d["friendly_name"])
        print("Serial Number   :", d["serial"])
        print("First Connected :", d["first_connected"])
        print("Last Connected  :", d["last_connected"])
        print()


# ----------------------------------------------------------
# Run the Test
# ----------------------------------------------------------

test_experiment2()
```

```python
# ==========================================================
# EXPERIMENT - 3
# Experiment Name: Linux SSH Auth Log Suspicious Login Detector
# Topic: Linux Forensics (/var/log/auth.log)
#
# Aim:
# Parse simplified Linux SSH authentication log lines and
# detect suspicious successful logins from untrusted IP
# addresses, with HIGH severity for root account logins.
# ==========================================================

import re

# ----------------------------------------------------------
# Regular Expression to Parse SSH Authentication Log Entries
# ----------------------------------------------------------

AUTH_LINE_RE = re.compile(
    r"(?P<result>Accepted|Failed) password for (?P<user>\S+) "
    r"from (?P<ip>[\d.]+) port (?P<port>\d+)"
)

# ----------------------------------------------------------
# Function to Parse auth.log Entries
# ----------------------------------------------------------

def parse_auth_log(lines):
    """
    Parse raw auth.log lines into structured dictionaries.
    """

    entries = []

    for line in lines:
        match = AUTH_LINE_RE.search(line)

        if match:
            entries.append({
                "result": match.group("result"),
                "user": match.group("user"),
                "ip": match.group("ip"),
                "port": int(match.group("port")),
                "raw": line,
            })

    return entries


# ----------------------------------------------------------
# Function to Detect Suspicious Logins
# ----------------------------------------------------------

def flag_suspicious_logins(entries, trusted_ips):
    """
    Flag successful logins from untrusted IP addresses.
    HIGH severity for root account.
    """

    flagged = []

    for e in entries:
        if e["result"] == "Accepted" and e["ip"] not in trusted_ips:

            severity = "HIGH" if e["user"] == "root" else "MEDIUM"

            flagged.append({
                **e,
                "severity": severity
            })

    return flagged


# ----------------------------------------------------------
# Simulated Linux SSH Authentication Log
# ----------------------------------------------------------

log_lines = [
    "Jan 15 08:00:01 server sshd[1001]: Accepted password for deploy from 10.0.0.5 port 51100 ssh2",
    "Jan 15 03:12:01 server sshd[1233]: Failed password for root from 198.51.100.23 port 51320 ssh2",
    "Jan 15 03:12:05 server sshd[1234]: Accepted password for root from 198.51.100.23 port 51322 ssh2",
]

trusted_ips = {"10.0.0.5"}


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment3():

    entries = parse_auth_log(log_lines)

    assert len(entries) == 3

    flagged = flag_suspicious_logins(entries, trusted_ips)

    assert len(flagged) == 1
    assert flagged[0]["user"] == "root"
    assert flagged[0]["ip"] == "198.51.100.23"
    assert flagged[0]["severity"] == "HIGH"

    print("All test cases passed.\n")

    print("===== Suspicious Login Report =====")

    for login in flagged:
        print("User      :", login["user"])
        print("Result    :", login["result"])
        print("IP Address:", login["ip"])
        print("Port      :", login["port"])
        print("Severity  :", login["severity"])
        print()


# ----------------------------------------------------------
# Run the Test
# ----------------------------------------------------------

test_experiment3()
```

```python
# ==========================================================
# EXPERIMENT - 4
# Experiment Name: File Timestamp (MACB) Timestomping Detector
# Topic: Windows / Linux Forensics (NTFS $MFT & inode timestamps)
#
# Aim:
# Detect timestomping (anti-forensics) by identifying
# impossible or suspicious timestamp relationships in a file's
# MACB (Modified, Accessed, Changed, Born) metadata.
# ==========================================================

from datetime import datetime

# ----------------------------------------------------------
# Timestamp Format
# ----------------------------------------------------------

TS_FMT = "%Y-%m-%d %H:%M:%S"

# ----------------------------------------------------------
# Function to Detect Timestomping
# ----------------------------------------------------------

def detect_timestomping(file_meta, change_gap_minutes=60):
    """
    file_meta:
    {
        "modified",
        "accessed",
        "changed",
        "born"
    }

    Returns:
    (is_suspicious, reasons)
    """

    m = datetime.strptime(file_meta["modified"], TS_FMT)
    a = datetime.strptime(file_meta["accessed"], TS_FMT)
    c = datetime.strptime(file_meta["changed"], TS_FMT)
    b = datetime.strptime(file_meta["born"], TS_FMT)

    reasons = []

    # Modified time cannot be before file creation
    if m < b:
        reasons.append("Modified time is earlier than Born (creation) time.")

    # Accessed time cannot be before file creation
    if a < b:
        reasons.append("Accessed time is earlier than Born (creation) time.")

    # Check for suspicious gap between Changed and Modified
    gap_minutes = abs((c - m).total_seconds()) / 60

    if gap_minutes > change_gap_minutes and c > m:
        reasons.append(
            f"MFT Changed time is {gap_minutes:.0f} minutes after Modified time. "
            "Metadata may have been altered after the fact."
        )

    return (len(reasons) > 0, reasons)


# ----------------------------------------------------------
# Simulated File Metadata
# ----------------------------------------------------------

normal_file = {
    "born": "2026-01-10 09:00:00",
    "modified": "2026-01-10 09:05:00",
    "accessed": "2026-01-12 14:00:00",
    "changed": "2026-01-10 09:05:00",
}

tampered_file = {
    "born": "2026-02-01 12:00:00",
    "modified": "2020-01-01 00:00:00",
    "accessed": "2026-02-01 12:00:00",
    "changed": "2026-02-01 12:03:00",
}


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment4():

    # Test normal file
    is_susp1, reasons1 = detect_timestomping(normal_file)
    assert is_susp1 is False

    # Test tampered file
    is_susp2, reasons2 = detect_timestomping(tampered_file)
    assert is_susp2 is True
    assert any("Modified time is earlier" in r for r in reasons2)

    print("All test cases passed.\n")

    print("===== File Timestamp Analysis =====")

    print("\nNormal File")
    print("Suspicious :", is_susp1)
    if reasons1:
        for reason in reasons1:
            print("-", reason)
    else:
        print("No suspicious timestamp relationships found.")

    print("\nTampered File")
    print("Suspicious :", is_susp2)
    print("Reasons:")
    for reason in reasons2:
        print("-", reason)


# ----------------------------------------------------------
# Run the Test
# ----------------------------------------------------------

test_experiment4()
```

```python
# ==========================================================
# EXPERIMENT - 5
# Experiment Name: Port Scan Detector from Packet Metadata
# Topic: Packet Capture and Traffic Analysis
#
# Aim:
# Analyze simulated packet records to detect a port scan by
# identifying a source IP that probes many destination ports
# on the same target within a short time window.
# ==========================================================

from datetime import datetime, timedelta

# ----------------------------------------------------------
# Timestamp Format
# ----------------------------------------------------------

PKT_FMT = "%Y-%m-%d %H:%M:%S"

# ----------------------------------------------------------
# Function to Detect Port Scans
# ----------------------------------------------------------

def detect_port_scan(packets, port_threshold=10, window_seconds=30):
    """
    Detect (src_ip -> dst_ip) pairs that contact
    port_threshold or more distinct destination ports
    within window_seconds.
    """

    # Sort packets by timestamp
    packets = sorted(
        packets,
        key=lambda p: datetime.strptime(p["timestamp"], PKT_FMT)
    )

    # Group packets by (Source IP, Destination IP)
    by_pair = {}

    for p in packets:
        key = (p["src_ip"], p["dst_ip"])
        by_pair.setdefault(key, []).append(p)

    results = {}

    # Check each communication pair
    for key, pkts in by_pair.items():

        for i in range(len(pkts)):

            start = datetime.strptime(pkts[i]["timestamp"], PKT_FMT)
            end = start + timedelta(seconds=window_seconds)

            ports_in_window = {
                p["dst_port"]
                for p in pkts
                if start <= datetime.strptime(p["timestamp"], PKT_FMT) <= end
            }

            if len(ports_in_window) >= port_threshold:
                results[key] = {
                    "distinct_ports": len(ports_in_window),
                    "ports": sorted(ports_in_window),
                }
                break

    return results


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment5():

    packets = []

    base = datetime(2026, 3, 1, 10, 0, 0)

    # Simulated Port Scan
    for i, port in enumerate(range(20, 32)):
        packets.append({
            "src_ip": "203.0.113.99",
            "dst_ip": "10.0.0.10",
            "dst_port": port,
            "timestamp": (base + timedelta(seconds=2 * i)).strftime(PKT_FMT),
        })

    # Normal HTTPS Traffic
    for i in range(5):
        packets.append({
            "src_ip": "10.0.0.20",
            "dst_ip": "10.0.0.30",
            "dst_port": 443,
            "timestamp": (base + timedelta(seconds=5 * i)).strftime(PKT_FMT),
        })

    results = detect_port_scan(
        packets,
        port_threshold=10,
        window_seconds=30
    )

    assert ("203.0.113.99", "10.0.0.10") in results
    assert results[("203.0.113.99", "10.0.0.10")]["distinct_ports"] >= 10
    assert ("10.0.0.20", "10.0.0.30") not in results

    print("All test cases passed.\n")

    print("===== Port Scan Detection Report =====")

    for (src, dst), info in results.items():
        print("Source IP         :", src)
        print("Destination IP    :", dst)
        print("Distinct Ports    :", info["distinct_ports"])
        print("Ports Scanned     :", info["ports"])
        print()


# ----------------------------------------------------------
# Run the Test
# ----------------------------------------------------------

test_experiment5()
```

```python
# ==========================================================
# EXPERIMENT - 6
# Experiment Name: DNS Tunneling Detector
# Topic: Packet Capture and Traffic Analysis / Intrusion Detection Basics
#
# Aim:
# Identify possible DNS tunneling by detecting DNS queries
# containing unusually long and high-entropy subdomains,
# which are commonly used for covert data exfiltration or
# command-and-control (C2) communication.
# ==========================================================

import math
from collections import Counter

# ----------------------------------------------------------
# Function to Calculate Shannon Entropy
# ----------------------------------------------------------

def shannon_entropy(s):
    """
    Calculate the Shannon entropy of a string.
    Higher entropy indicates more random-looking data.
    """

    if not s:
        return 0.0

    counts = Counter(s)
    length = len(s)

    return -sum(
        (count / length) * math.log2(count / length)
        for count in counts.values()
    )


# ----------------------------------------------------------
# Function to Detect DNS Tunneling
# ----------------------------------------------------------

def detect_dns_tunneling(
    queries,
    length_threshold=20,
    entropy_threshold=3.5
):
    """
    Detect DNS tunneling by checking whether the leftmost
    subdomain is both long and has high Shannon entropy.
    """

    flagged = []

    for q in queries:

        # Extract leftmost label
        label = q.split(".")[0]

        entropy = shannon_entropy(label)

        if (
            len(label) >= length_threshold
            and entropy >= entropy_threshold
        ):
            flagged.append({
                "query": q,
                "label_length": len(label),
                "entropy": round(entropy, 2)
            })

    return flagged


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment6():

    queries = [
        "www.google.com",
        "mail.office365.com",
        "a8f3k2j9x1p7q4z6w0n5r2t8y3.exfil-domain.com",
        "vpn.corporate-network.com",
        "9c2e7b1a4f8d3c6e0a5b9d2f7c1e4a8b3d6f9c2e5a8b1d4f.tunnel.example.net",
    ]

    flagged = detect_dns_tunneling(queries)

    flagged_domains = [f["query"] for f in flagged]

    assert "www.google.com" not in flagged_domains
    assert "mail.office365.com" not in flagged_domains
    assert "vpn.corporate-network.com" not in flagged_domains

    assert "a8f3k2j9x1p7q4z6w0n5r2t8y3.exfil-domain.com" in flagged_domains

    assert "9c2e7b1a4f8d3c6e0a5b9d2f7c1e4a8b3d6f9c2e5a8b1d4f.tunnel.example.net" in flagged_domains

    assert len(flagged) == 2

    print("All test cases passed.\n")

    print("===== DNS Tunneling Detection Report =====")

    for item in flagged:
        print("Query           :", item["query"])
        print("Label Length    :", item["label_length"])
        print("Entropy         :", item["entropy"])
        print()


# ----------------------------------------------------------
# Run the Test
# ----------------------------------------------------------

test_experiment6()
```

```python
# ==========================================================
# EXPERIMENT - 7
# Experiment Name: Signature-Based IDS Packet Payload Matcher
# Topic: Intrusion Detection Basics
#
# Aim:
# Build a simple signature-based Intrusion Detection System
# (IDS) that scans packet payloads for known malicious
# patterns and generates alerts.
# ==========================================================

# ----------------------------------------------------------
# Signature Database
# ----------------------------------------------------------

SIGNATURES = [
    {
        "sid": 1000001,
        "name": "Possible SQL Injection",
        "pattern": "union select",
    },
    {
        "sid": 1000002,
        "name": "Directory Traversal Attempt",
        "pattern": "../../../etc/passwd",
    },
    {
        "sid": 1000003,
        "name": "Nmap Scripting Engine User-Agent",
        "pattern": "nmap scripting engine",
    },
]

# ----------------------------------------------------------
# Function to Scan Packet Payloads
# ----------------------------------------------------------

def scan_payloads(packets, signatures=SIGNATURES):
    """
    Scan packet payloads for known attack signatures.

    Parameters:
        packets : List of packet dictionaries containing
                  a 'payload' field.

    Returns:
        List of IDS alerts.
    """

    alerts = []

    for idx, pkt in enumerate(packets):

        payload_lower = pkt["payload"].lower()

        for sig in signatures:

            if sig["pattern"] in payload_lower:

                alerts.append({
                    "packet_index": idx,
                    "sid": sig["sid"],
                    "name": sig["name"],
                })

    return alerts


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment7():

    packets = [
        {
            "payload": "GET /products?id=1 HTTP/1.1"
        },
        {
            "payload": "GET /login?user=admin' UNION SELECT username,password FROM users-- HTTP/1.1"
        },
        {
            "payload": "GET /download?file=../../../etc/passwd HTTP/1.1"
        },
    ]

    alerts = scan_payloads(packets)

    alert_indices = {a["packet_index"] for a in alerts}

    assert 0 not in alert_indices
    assert 1 in alert_indices
    assert 2 in alert_indices

    assert any(
        a["name"] == "Possible SQL Injection"
        for a in alerts
    )

    assert any(
        a["name"] == "Directory Traversal Attempt"
        for a in alerts
    )

    print("All test cases passed.\n")

    print("===== IDS Alert Report =====")

    for alert in alerts:
        print("Packet Index :", alert["packet_index"])
        print("Signature ID :", alert["sid"])
        print("Alert Name   :", alert["name"])
        print()


# ----------------------------------------------------------
# Run the Test
# ----------------------------------------------------------

test_experiment7()
```

```python
# ==========================================================
# EXPERIMENT - 9
# Experiment Name: Cloud Audit Log Anomalous Download Detector
# Topic: Cloud Forensics Fundamentals
#
# Aim:
# Analyze simulated cloud storage audit logs to detect
# anomalous file download activity based on unfamiliar IP
# addresses or downloads occurring outside normal
# business hours.
# ==========================================================

from datetime import datetime
from collections import Counter

# ----------------------------------------------------------
# Timestamp Format
# ----------------------------------------------------------

LOG_FMT = "%Y-%m-%d %H:%M:%S"

# ----------------------------------------------------------
# Function to Build Baseline IP for Each User
# ----------------------------------------------------------

def build_baseline_ips(logs):
    """
    Returns a dictionary containing the most common IP
    address used by each user.
    """

    by_user = {}

    for entry in logs:
        by_user.setdefault(entry["user"], []).append(entry["ip"])

    return {
        user: Counter(ips).most_common(1)[0][0]
        for user, ips in by_user.items()
    }


# ----------------------------------------------------------
# Function to Detect Anomalous Downloads
# ----------------------------------------------------------

def flag_anomalous_downloads(logs, business_start=8, business_end=20):
    """
    Flags download events that occur:
    1. From an unfamiliar IP address.
    2. Outside business hours (08:00 - 20:00).
    """

    baseline = build_baseline_ips(logs)

    flagged = []

    for entry in logs:

        if entry["action"] != "download":
            continue

        ts = datetime.strptime(entry["timestamp"], LOG_FMT)

        reasons = []

        if entry["ip"] != baseline.get(entry["user"]):
            reasons.append("IP differs from user baseline")

        if not (business_start <= ts.hour < business_end):
            reasons.append("Outside business hours")

        if reasons:
            flagged.append({
                **entry,
                "reasons": reasons
            })

    return flagged


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment9():

    logs = [
        {
            "user": "alice",
            "action": "view",
            "file": "roadmap.docx",
            "timestamp": "2026-03-01 10:00:00",
            "ip": "10.0.0.5"
        },
        {
            "user": "alice",
            "action": "view",
            "file": "budget.xlsx",
            "timestamp": "2026-03-02 11:00:00",
            "ip": "10.0.0.5"
        },
        {
            "user": "alice",
            "action": "download",
            "file": "report.pdf",
            "timestamp": "2026-03-03 14:00:00",
            "ip": "10.0.0.5"
        },
        {
            "user": "alice",
            "action": "download",
            "file": "customer_database.csv",
            "timestamp": "2026-03-05 02:15:00",
            "ip": "185.220.101.7"
        }
    ]

    flagged = flag_anomalous_downloads(logs)

    assert len(flagged) == 1
    assert flagged[0]["file"] == "customer_database.csv"
    assert "IP differs from user baseline" in flagged[0]["reasons"]
    assert "Outside business hours" in flagged[0]["reasons"]

    print("All test cases passed.\n")

    print("===== Cloud Audit Log Anomaly Report =====")

    for event in flagged:
        print("User       :", event["user"])
        print("Action     :", event["action"])
        print("File       :", event["file"])
        print("Timestamp  :", event["timestamp"])
        print("IP Address :", event["ip"])
        print("Reasons    :", ", ".join(event["reasons"]))
        print()


# ----------------------------------------------------------
# Run the Test
# ----------------------------------------------------------

test_experiment9()
```

```python
# ==========================================================
# EXPERIMENT - 10
# Experiment Name: Malware Static Analysis - Hashing and Entropy Check
# Topic: Malware Analysis Basics
#
# Aim:
# Perform static malware analysis by computing the SHA-256
# hash of a file and calculating its Shannon entropy to
# identify whether the file is likely packed or encrypted.
# ==========================================================

import hashlib
import math
import os
from collections import Counter

# ----------------------------------------------------------
# Function to Compute SHA-256 Hash
# ----------------------------------------------------------

def sha256_hash(data: bytes) -> str:
    """
    Returns the SHA-256 hash of the given byte data.
    """
    return hashlib.sha256(data).hexdigest()


# ----------------------------------------------------------
# Function to Calculate Byte-Level Shannon Entropy
# ----------------------------------------------------------

def byte_entropy(data: bytes) -> float:
    """
    Calculates the Shannon entropy of byte data.
    Higher entropy usually indicates compressed,
    packed, or encrypted content.
    """

    if not data:
        return 0.0

    counts = Counter(data)
    length = len(data)

    return -sum(
        (count / length) * math.log2(count / length)
        for count in counts.values()
    )


# ----------------------------------------------------------
# Function to Perform Static Analysis
# ----------------------------------------------------------

def static_analyze(data: bytes, entropy_threshold=7.5):
    """
    Performs static analysis on byte data.

    Returns:
    - SHA-256 Hash
    - File Size
    - Shannon Entropy
    - Likely Packed (True/False)
    """

    entropy = byte_entropy(data)

    return {
        "sha256": sha256_hash(data),
        "size_bytes": len(data),
        "entropy": round(entropy, 2),
        "likely_packed": entropy >= entropy_threshold,
    }


# ----------------------------------------------------------
# Test Cases
# ----------------------------------------------------------

def test_experiment10():

    # Low-entropy sample (plain text)
    plain_sample = (
        b"This is a normal configuration file. " * 20
    )

    # High-entropy sample (random bytes)
    random_sample = os.urandom(2000)

    plain_result = static_analyze(plain_sample)
    random_result = static_analyze(random_sample)

    assert len(plain_result["sha256"]) == 64
    assert plain_result["entropy"] < 5.0
    assert plain_result["likely_packed"] is False

    assert random_result["entropy"] >= 7.5
    assert random_result["likely_packed"] is True

    assert sha256_hash(plain_sample) == sha256_hash(plain_sample)

    print("All test cases passed.\n")

    print("===== Malware Static Analysis Report =====")

    print("\nPlain Sample")
    print("SHA-256 Hash :", plain_result["sha256"])
    print("File Size    :", plain_result["size_bytes"], "bytes")
    print("Entropy      :", plain_result["entropy"])
    print("Likely Packed:", plain_result["likely_packed"])

    print("\nRandom Sample")
    print("SHA-256 Hash :", random_result["sha256"])
    print("File Size    :", random_result["size_bytes"], "bytes")
    print("Entropy      :", random_result["entropy"])
    print("Likely Packed:", random_result["likely_packed"])


# ----------------------------------------------------------
# Run the Test
# ----------------------------------------------------------

test_experiment10()
```
