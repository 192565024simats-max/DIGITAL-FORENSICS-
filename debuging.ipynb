```python
def classify_incident(description):
    categories = {
        'Phishing': ['fake email', 'login link', 'credential'],
        'Ransomware': ['encrypted files', 'ransom note', 'bitcoin'],
        'Identity Theft': ['stolen identity', 'aadhaar', 'pan card']
    }

    description = description.lower()

    for category, keywords in categories.items():
        for word in keywords:
            if word in description:
                return category

    return 'Unclassified'


result = classify_incident('Victim received a fake email with a login link')
print(result)
```

```python
custody_log = []

def add_entry(evidence_id, handler, action):
    entry = {
        'evidence_id': evidence_id,
        'handler': handler,
        'action': action
    }
    custody_log.append(entry)

def print_log():
    for entry in custody_log:
        print(entry['evidence_id'] + " - " + entry['handler'] + " - " + entry['action'])

add_entry('EVD-101', 'A.Kumar', 'Collected')
add_entry('EVD-101', 'S.Rao', 'Analysed')

print_log()
```

```python
import re

def is_suspicious(url):
    ip_pattern = r'\d+\.\d+\.\d+\.\d+'
    suspicious_words = ('verify', 'update', 'secure', 'login')

    # Check if the URL contains an IP address
    if re.search(ip_pattern, url):
        return True

    # Check if the URL contains any suspicious word
    for word in suspicious_words:
        if word in url:
            return True

    return False

urls = [
    'http://192.168.1.10/verify-account',
    'https://mybank.com/login'
]

for u in urls:
    print(u, '->', is_suspicious(u))
```

```python
def detect_bruteforce(log_lines, threshold=5):
    ip_counts = {}

    for line in log_lines:
        if 'FAILED LOGIN' in line:
            parts = line.split(' ')
            ip = parts[-1]

            if ip in ip_counts:
                ip_counts[ip] += 1
            else:
                ip_counts[ip] = 1

    flagged = []

    for ip, count in ip_counts.items():
        if count > threshold:
            flagged.append(ip)

    return flagged


log = [
    'FAILED LOGIN user=admin from 10.0.0.5',
    'FAILED LOGIN user=admin from 10.0.0.5',
    'FAILED LOGIN user=root from 10.0.0.5'
]

print(detect_bruteforce(log, threshold=2))
```
